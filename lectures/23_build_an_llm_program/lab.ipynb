## Deepening: derive GPU count from the workload

This section turns the notebook into a quantitative lab. Predict before executing.


```python
import math

def dense_flops(params, tokens):
    return 6 * params * tokens

def gpu_count(required_flops, wall_seconds, gpu_flops, efficiency):
    return required_flops / (wall_seconds * gpu_flops * efficiency)

params = 7e9
tokens = 1e11
F = dense_flops(params, tokens)
print("planning FLOPs:", F)

for efficiency in [0.9, 0.7, 0.5]:
    g = gpu_count(F, 7*24*3600, 5e15, efficiency)
    print("efficiency", efficiency, "estimated GPUs:", math.ceil(g))
```
