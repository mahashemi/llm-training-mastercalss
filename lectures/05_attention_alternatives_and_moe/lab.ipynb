## Deepening: head geometry and routing

This section upgrades the notebook from a runnable demo to a measurable experiment. Predict before executing.


```python
import torch
print("torch:", torch.__version__)
print("cuda:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("gpu:", torch.cuda.get_device_name(0))
```

```python
import torch, time

device = "cuda" if torch.cuda.is_available() else "cpu"
print("device:", device)

def simple_attention(q, k, v):
    scores = q @ k.transpose(-2, -1)
    weights = torch.softmax(scores / (q.shape[-1] ** 0.5), dim=-1)
    return weights @ v

B, L, Hq, Hkv, d = 2, 512, 16, 4, 64
q = torch.randn(B, Hq, L, d, device=device, dtype=torch.float16 if device=="cuda" else torch.float32)
k = torch.randn(B, Hkv, L, d, device=q.dtype)
v = torch.randn(B, Hkv, L, d, device=q.dtype)
repeat_factor = Hq // Hkv
kg = k.repeat_interleave(repeat_factor, dim=1)
vg = v.repeat_interleave(repeat_factor, dim=1)

if device == "cuda": torch.cuda.reset_peak_memory_stats()
t0 = time.perf_counter()
y = simple_attention(q, kg, vg)
if device == "cuda": torch.cuda.synchronize()
print("output shape:", tuple(y.shape))
print("elapsed seconds:", time.perf_counter() - t0)
if device == "cuda": print("peak GiB:", torch.cuda.max_memory_allocated()/2**30)
```

## 7. Research exercise — real workload → KV-cache design

Use real FineWeb document lengths to create a serving workload. Compare MHA, GQA, and MQA KV-cache footprints for the same model dimensions. Report memory per request and aggregate memory at concurrency 1, 8, and 32.


```python
from datasets import load_dataset
import numpy as np

ds=load_dataset("HuggingFaceFW/fineweb","sample-10BT",split="train",streaming=True)
lengths=[]
for row in ds:
    t=row.get("text","")
    if t:
        lengths.append(min(4096,max(128,len(t)//4)))
    if len(lengths)>=200: break

layers,heads,head_dim,bytes_per=32,32,128,2
for kv_heads in [32,8,1]:
    per_token=2*layers*kv_heads*head_dim*bytes_per
    for concurrency in [1,8,32]:
        p95=np.percentile(lengths,95)
        gib=per_token*p95*concurrency/2**30
        print(f"KV heads={kv_heads:2d} concurrency={concurrency:2d} p95_context={p95:.0f} approx_KV_GiB={gib:.3f}")
```

<details><summary><strong>Expected findings / answer</strong></summary>

With all other dimensions fixed, KV memory is approximately proportional to the number of KV heads. GQA therefore reduces cache pressure relative to MHA, and MQA reduces it further. The trade-off is that fewer independent K/V heads can affect model quality and attention behavior. The experiment teaches the systems decision: **use GQA/MQA when KV-cache memory is actually a bottleneck, and validate quality rather than treating cache reduction as free.**
</details>
