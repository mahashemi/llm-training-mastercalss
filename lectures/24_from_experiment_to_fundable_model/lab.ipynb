## 7. Research exercise — convert real corpus evidence into a scale plan


```python
from datasets import load_dataset
from collections import Counter

sources=[
 ("FineWeb","HuggingFaceFW/fineweb","sample-10BT"),
 ("FineWeb-Edu","HuggingFaceFW/fineweb-edu","sample-10BT"),
 ("OpenWebMath","open-web-math/open-web-math",None),
]
evidence=[]
for label,ds_id,config in sources:
    kwargs={"split":"train","streaming":True}
    if config: kwargs["name"]=config
    try:
        ds=load_dataset(ds_id,**kwargs)
        rows=0; chars=0
        for row in ds:
            t=str(row.get("text",""))
            if t: rows+=1; chars+=len(t)
            if rows>=100: break
        evidence.append((label,rows,chars))
    except Exception as e:
        evidence.append((label,"SKIP",str(e)[:80]))
print("source evidence:")
for x in evidence: print(x)

# A transparent scaling budget.
N=1e9
D=300e9
flops=6*N*D
print("1B params × 300B tokens -> approx FLOPs:",f"{flops:.3e}")
```
