# Lab 23 — National/Organizational LLM Resource Planner
**Goal:** Translate a model target into first-order compute and a transparent funding scenario.

Free-first: CPU or free Colab where possible; expensive paths are clearly marked optional.



```python
target_params=7e9
tokens=140e9
flops=6*target_params*tokens
print("rough training FLOPs:",flops/1e21,"ZFLOPs")
```

```python
gpu_peak_tflops=1000
utilization=0.30
gpu_hours=flops/(gpu_peak_tflops*1e12*3600*utilization)
print("illustrative GPU-hours:",gpu_hours)
```

```python
hourly_price=5.0
print("illustrative compute cost (USD):",gpu_hours*hourly_price)
```

### Answer
Treat every planning number as an assumption until validated against measured throughput, current vendor quotes, network/storage architecture, checkpoint overhead, staffing, and failure/retry rates. A credible proposal makes uncertainty explicit and uses pilots to reduce it.
