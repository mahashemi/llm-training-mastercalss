### Dataset
We start with a deliberately tiny dataset so every example is inspectable. In a serious run, dataset design and quality must be established before scaling.


```python
from datasets import Dataset
data=Dataset.from_list([
 {"prompt":"What is a GPU?","completion":"A GPU is a processor designed for highly parallel numerical computation."},
 {"prompt":"Explain overfitting simply.","completion":"Overfitting happens when a model learns the training examples too closely and performs poorly on new examples."},
 {"prompt":"Why use a validation set?","completion":"To estimate performance on unseen examples while choosing the training configuration."},
])
print(data)
```

```python
# Optional GPU path.
from trl import SFTConfig, SFTTrainer
args=SFTConfig(
    output_dir="outputs/sft-demo",
    learning_rate=2e-5,
    per_device_train_batch_size=2,
    num_train_epochs=1,
    max_length=256,
    logging_steps=1,
    save_strategy="no",
    report_to=[]
)
trainer=SFTTrainer(model="Qwen/Qwen3-0.6B", args=args, train_dataset=data)
trainer.train()
```

### Critical thinking
