# Lab 02 — Tokenizer Design and Measurement
**Goal:** Compare character, whitespace, and a simple BPE-style tokenizer. Measure vocabulary size and token fertility, then discuss multilingual implications.

Free-first: run on CPU or free Colab where possible.

Method: **predict → run → inspect → break → decide**.


```python
text = "The tokenizer determines what the model can represent efficiently. भारत India فارسی Persian العربية Arabic."
chars = list(text)
words = text.split()
print("characters:", len(chars))
print("whitespace words:", len(words))
```

## Predict before running
Predict:
- Which representation will have the largest sequence length?
- What happens when a single word is split into many pieces?
- Which matters more for a language program: raw vocabulary count or tokens per semantic unit?


```python
import re
def rough_subwords(s):
    return re.findall(r"[A-Za-z]+|\d+|[^\w\s]", s, flags=re.UNICODE)
subs = rough_subwords(text)
print("rough subword-like pieces:", len(subs))
print("fertility char/word:", len(chars)/max(1,len(words)))
print("fertility rough/word:", len(subs)/max(1,len(words)))
```

## Explain the result
The experiment should make clear that tokenizer choice changes sequence length and therefore affects compute, context utilization, memory, and coverage. A multilingual tokenizer should be evaluated on representative text from every target language rather than optimized only for English.


```python
# Try three strings: English, Persian, and a language you know well.
# Predict which gets the highest token count before measuring.
examples = [
    "The patient returned home after surgery.",
    "بیمار پس از عمل جراحی به خانه بازگشت.",
    "रोगी सर्जरी के बाद घर लौट आया।",
]
for x in examples:
    print(x)
    print("chars:", len(x), "pieces:", len(rough_subwords(x)))
```

## Critical-thinking answers
A tokenizer is part of the model/data interface. High fertility can inflate training and inference cost; low-resource scripts can also suffer from poor representation. The right objective is not simply “smallest vocabulary” or “fewest tokens,” but a balanced tradeoff across languages, domains, training data, memory, and downstream quality.
