```python
import torch
print("torch:", torch.__version__)
print("cuda:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("gpu:", torch.cuda.get_device_name(0))
```

```python
from dataclasses import dataclass, asdict

@dataclass
class ExperimentGate:
    question: str
    hypothesis: str
    intervention: str
    metric: str
    resource_metric: str
    success_threshold: str
    kill_criterion: str

gate = ExperimentGate(
    "Does PEFT close the measured behavior gap?",
    "Representative SFT improves target behavior without unacceptable regression.",
    "LoRA rank 16 on a small open-weight model",
    "target task score",
    "peak GPU memory + tokens/sec",
    "predeclared target threshold",
    "target gain below minimum after agreed budget",
)
print(asdict(gate))
```
