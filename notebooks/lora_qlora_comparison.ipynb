# Lab 18 — LoRA and QLoRA Comparison
**Goal:** Calculate low-rank adapter size and inspect a current PEFT QLoRA configuration.

Free-first: CPU or free Colab where possible; expensive paths are clearly marked optional.



```python
def lora_params(in_dim,out_dim,r): return r*(in_dim+out_dim)
full=4096*4096
for r in [4,8,16,32,64]:
    p=lora_params(4096,4096,r)
    print(r,p, "ratio to full:",p/full)
```

```python
import torch
from transformers import BitsAndBytesConfig
if torch.cuda.is_available():
    qconfig=BitsAndBytesConfig(load_in_4bit=True,bnb_4bit_quant_type="nf4",
                               bnb_4bit_use_double_quant=True,
                               bnb_4bit_compute_dtype=torch.bfloat16)
    print(qconfig)
```

```python
from peft import LoraConfig
config=LoraConfig(r=16,lora_alpha=32,lora_dropout=0.05,target_modules="all-linear",
                  task_type="CAUSAL_LM")
print(config)
```

### Answer
LoRA freezes the base model and learns low-rank updates. QLoRA adds low-bit quantization of the base model to reduce memory. These techniques can make adapter training practical on much smaller hardware than full fine-tuning. Evaluate quality and failure modes rather than assuming equivalence.
